```python
#What age groups are the best in the 50mi race (20+ races min)

data2.query("race_length == '50mi'") \
   .groupby('Athlete_age')['athlete_average_speed'] \
   .agg(['mean', 'count']) \
   .sort_values('mean', ascending=False) \
   .query('count > 19') \
   .head(15)
```

```python
# Convert the index (athlete_age) to a list
age_list = data2.query("race_length == '50mi'") \
   .groupby('Athlete_age')['athlete_average_speed'] \
   .agg(['mean', 'count']) \
   .sort_values('mean', ascending=False) \
   .query('count > 19') \
    .index \
    .tolist()

# Find the position of age 49
age_to_find = 49
if age_to_find in age_list:
    position = age_list.index(age_to_find) + 1  # Add 1 to convert to 1-based indexing
    print(f"Age {age_to_find} is in position {position} in total of {len(age_list)}.")
else:
    print(f"Age {age_to_find} is not in the sorted data.")
```

```python
#Are runners slower in summer than winter?
data2['race_month'] = data2['race_day'].str.split('.').str.get(1).astype(int)
```

```python
data2['race_season'] = data2['race_month'].apply(lambda x: 'Winter' if x > 11 else 'Fall' if x > 8 else 'Summer' if x > 5 else 'Spring' if x > 2 else 'Winter')
```

```python
data2.groupby('race_season')['athlete_average_speed'].agg(['mean', 'count']).sort_values('mean', ascending=False)
```

```python
#We can see from the data that the runners are most probably slower in summer than winter.
# 50mi only
data2.query("race_length == '50mi'").groupby('race_season')['athlete_average_speed'].agg(['mean', 'count']).sort_values('mean', ascending=False)
```
